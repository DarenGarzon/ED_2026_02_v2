```python
def find_min(T:LinkedBinaryTree):
    '''Retorna el contenedor donde se encuentra el mínimo'''
    valor = T.root()
        
    while T.left(valor) is not None:
        valor = T.left(valor)
        
    return valor
```

```python
find_min(L).element()
```

```python
def es_ordenado(T:LinkedBinaryTree):
    I = T.inorder()
    while True:
        try:
            v1 = next(I)
            v2 = next(I)
            if v1 > v2:
                return False
        except:
            return True
```
